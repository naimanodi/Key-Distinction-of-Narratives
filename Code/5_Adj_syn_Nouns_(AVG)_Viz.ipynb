```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

excel_file = "Adj_syn_Nouns_(AVG).xlsx"

df = pd.read_excel(excel_file)

#Extracts the Average row
avg_row = df[df["Key Distinction"] == "Average"].iloc[0]

# Average values for categories
categories = ["Adjective", "Synonyms", "Noun"]

fasttext_scores = [
    avg_row["Adj_FastText"],
    avg_row["Syn_FastText"],
    avg_row["Noun_FastText"]
]

spacy_scores = [
    avg_row["Adj_SpaCy"],
    avg_row["Syn_SpaCy"],
    avg_row["Noun_SpaCy"]
]


x = np.arange(len(categories))
width = 0.35

plt.figure(figsize=(8, 5))

plt.bar(
    x - width/2,
    fasttext_scores,
    width,
    label="FastText"
)

plt.bar(
    x + width/2,
    spacy_scores,
    width,
    label="SpaCy"
)

for i, v in enumerate(fasttext_scores):
    plt.text(i - width/2, v + 0.01, f"{v:.3f}", ha="center")

for i, v in enumerate(spacy_scores):
    plt.text(i + width/2, v + 0.01, f"{v:.3f}", ha="center")

plt.xticks(x, categories)
plt.ylabel("Average F1-Score")
plt.xlabel("Form of Representation")
plt.title("Average F1-Score Comparison: FastText vs SpaCy")
plt.ylim(0, 1)
plt.legend()
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()

#Saves the plot
plt.savefig(
    r"Result/average_f1_scores.png", #file location to save the plot
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
