```python
import pandas as pd
import krippendorff


file_path = "file_name_here.xlsx"
df = pd.read_excel(file_path)

annotator_columns = [
    "Annotator_1", #add no. of annotators according to the column name
    "Annotator_2",
    "etc..."
]

data = df[annotator_columns].to_numpy().T

alpha = krippendorff.alpha(data, level_of_measurement="ordinal") #level of measurement ordinal for our likert scale consideration

print(f"Krippendorff's Alpha (harmonious_disharmonious): {alpha:.3f}")
```

```python
key_distinction = [
    "harmonious_disharmonious", #all the key distinctions with their individual file names
    "natural_cultural",
    "etc..."
]

alpha_values = []

for path_kd in key_distinction:

    file_path = f"{path_kd}.xlsx"
    df = pd.read_excel(file_path)

    annotator_columns = [
        "Annotator_1", #add no. of annotators according to the column name
        "Annotator_2"
        "etc..."
    ]

    data = df[annotator_columns].to_numpy().T

    alpha = krippendorff.alpha(data, level_of_measurement="ordinal")

    alpha_values.append(alpha)

    print(f"Krippendorff's Alpha ({path_kd}): {alpha:.3f}")

#mean alpha
mean_alpha = sum(alpha_values) / len(alpha_values)

print(f"\n Mean Krippendorff's Alpha: {mean_alpha:.3f}")
```
